# Bepaling van soortelijke warmte van water


# Introductie

Water heeft een enorme buffercapaciteit: je moet veel energie toevoegen om de temperatuur van water een graad te verwarmen. In dit practicum gaan we de soortelijke warmte van water bepalen door een bekende hoeveelheid water te verwarmen met een bekende hoeveelheid energie, en de temperatuurstijging te meten.

# Theorie



### Warmtecapaciteit en soortelijke warmte
De warmtecapaciteit $C$ van een voorwerp is de hoeveelheid warmte $Q$ die nodig is om de temperatuur met 1 K te verhogen:

$$ C = \frac{Q}{\Delta T} \quad [\text{J/K}] $$

De soortelijke warmtecapaciteit $c$ is de warmtecapaciteit per massa-eenheid:

$$ c = \frac{C}{m} = \frac{Q}{m \Delta T} \quad [J/(kg \cdot K)] $$

waarbij $m$ de massa van de stof is.

De warmtecapaciteit is (volgens mij) een heel interessante grootheid. Een voorgaande proef was de bepaling van de constante van boltzmann. Deze constante speelt in de python simulaties (deel van dit portefolio) een belangrijke rol door deze vergelijking:

$$
    \frac{f}{2}kT=\frac{1}{2}m\left<v^2\right>
$$

waarbij $f$ het aantal vrijheidsgraden is. Deze vergelijking heeft duidelijk te maken met energie, waarbij deze formule voor een gasmodel een versimpeling is.

Een beetje chemie geeft inzicht in de geometrie van moleculen en de manieren waarop moleculen kunnen bewegen en interacties aangaan met anderen. Op deze manier worden extra vrijheidsgraden ontdekt, waardoor het evident wordt waarom water een hoge warmtecapaciteit heeft en waarom er verschillen zijn tussen verschillende stoffen op dit vlak.

Een watermolecuul ($H_2O$) bestaat uit een zuurstofatoom (O) gebonden aan 2 waterstofatomen (H) deze bindigen hebben een lengte en vormen een bindingshoek van $\approx 104 \degree $. Deze bindingen kunnen zich gedragen als veren. Ze oscilleren in verschillende configuraties, elk met een daarbijhorende totale energie van het syteem. Dit zal volstaan voor wat anngetoond moest worden: op verschillende niveaus naast translationele kan energie opgenomen worden. 

Warmtecapaciteit omvat dit alles, het vertelt je hoeveel energie je nodig hebt om een vergelijking, gelijkaardig aan bovenstaande, te voldoen voor een kilogram van je stof één graad optewarmen.

### Verwarming met constant vermogen
Bij een elektrisch verwarmingselement met constant vermogen $P$ geldt:

$$ Q = P t $$

In een ideaal geïsoleerd systeem leidt dit tot:

$$ P t = m c \Delta T \quad \Rightarrow \quad c = \frac{P t}{m \Delta T} \tag{1} $$



### Verwachting voor water
Voor water is de literatuurwaarde:

$$ c_{\text{water}} \approx 4186 \ J/(kg \cdot K) $$

In dit experiment wordt $c$ bepaald uit het lineaire deel van de temperatuur–tijdgrafiek na de initiële opwarmfase, gebruikmakend van vergelijking (1).

# Resultaten


````python
import numpy as np 
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
# Hier de data en de analyse

start_temp = 20.2 #C
time = np.arange(0, 181, 10)
temp = np.array([20.2, 20.2, 20.2, 20.2, 20.2, 20.2, 20.2, 20.3, 20.5, 20.8, 20.8, 21.0, 21.0, 21.0, 21.1, 21.2, 21.1, 21.3, 21.4])
````

````python
# Sla figuren op met  
# 
# plt.savefig("figures/naam.png", dpi=450)

P = 10**2/10 # W want P = U**2/R
c = P*180/((temp[-1]-start_temp) * 0.5) # op basis van het vermogen, totale tijd en temperatuursverschil

print(f'Op basis van het vermogen, temperattuursverschil en duur is de warmtecapaciteit {c:.2f} J/(K kg)')

def lin_fit(t,a,b):
    return a*t + b

# Het eerste deel van de data is plat dus we fitten enkel het stijgende deel

popt, cov = curve_fit(lin_fit, time[6:], temp[6:])

x = np.linspace(0, 180, 1000)

plt.figure()
plt.plot(time, temp, 'bo', label='meting')
plt.plot(x, lin_fit(x, *popt), 'r-', label='fit: a=%5.3f, b=%5.3f' % tuple(popt))
plt.xlabel('Tijd [s]')
plt.ylabel('Temperatuur [°C]')
plt.legend()
plt.grid()
plt.show()

a = popt[0]  # in K/s
# c_water = P/ (a*m)
c_water = P/ (a*0.5)  # in J/(K kg)
print(f'Op basis van de fit en het vermogen is de warmtecapaciteit van water: {c_water:.2f} J/(K kg)')
````
